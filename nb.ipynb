```python
#Problem 7
#Prime Number
number=29
is_prime=True
if number<=1:
    is_prime=False
else:
    for i in range(2,number):
        if number%i==0:
            is_prime=False
            break
print("Prime" if is_prime else "Not Prime")
```

```python
#Problem 11
#Print Star Pattern - Right Triangle
n = 5
for i in range(1, n + 1):
    print("*" * i)
```

```python
#Problem 12
#Print Star Pattern - Inverted Triangle
n = 5
for i in range(n, 0, -1):
    print("*" * i)
```

```python
#Problem 13
#Print Star Pattern - Pyramid
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
#Problem 22
#Find the union, intersection, and difference of two sets
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
print("Union:", sorted(set1 | set2))
print("Intersection:", sorted(set1 & set2))
print("Difference:", sorted(set1 - set2))
```
